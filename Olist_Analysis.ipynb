```python
import pandas as pd
```

```python
#Dataset 1 : Customer information with with their customer_id, customer_unique id and location
# customer_id links the customer record to a specific order.
# customer_unique_id identifies the same customer across multiple orders.
customer = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/olist_customers_dataset.csv")

#Dataset 2 : Geolocation information with latitute and longitude
geolocation = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/olist_geolocation_dataset.csv")

#Dataset 3 : Order items information with orders, products and sellers id with freight value and shipping limit date (carrier handoff date)
order_items = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/olist_order_items_dataset.csv")

#Dataset 4: Payment infomration with payment type and installements
order_payments = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/olist_order_payments_dataset.csv")

#Dataset 5: Reviews froms customers with score and comments
order_reviews = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/olist_order_reviews_dataset.csv")

#Dataset 6: Orders info with deliver and expected delivery dates
orders = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/olist_orders_dataset.csv")

#Dataset 7: Produt information with product catergory and its specifications
products = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/olist_products_dataset.csv")

#Dataset 8: Portuguese product name converted to english name
products_category = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/product_category_name_translation.csv")

#Dataset 9: Sellers information with their id and their location
sellers = pd.read_csv("/Volumes/workspace/default/olist_original_raw_files/olist_sellers_dataset.csv")
```

```python
#Dataset 1: Customer data profiling
display(customer.head())

print('Total Total no of rows and columns:', customer.shape)

display(
    customer.count()
    .to_frame('Non null count')
    .rename_axis('Column')
    .reset_index()
)

display(
    customer.nunique()
    .to_frame('Unique non null counts')
    .rename_axis('Column')
    .reset_index()
)

# 99,441 distinct customer_id values and 96,096 distinct customer_unique_id values.
# customer_id identifies an order-specific customer record.
# customer_unique_id identifies a customer across multiple orders.
# The difference indicates that some customers placed more than one order.
```

```python
#Dataset 2: Geolocation data profiling
display(geolocation.head())

print('Total no of rows and columns:', geolocation.shape)

display(
    geolocation.count()
    .to_frame('Non null count')
    .rename_axis('Column')
    .reset_index()
)

display(
    geolocation.nunique()
    .to_frame('Unique non null counts')
    .rename_axis('Column')
    .reset_index()
)

# Geolocation: 1,000,163 rows and 19,015 distinct ZIP prefixes.
# Multiple geographic records share the same ZIP prefix.
```

```python
#Dataset 3: Order items data profiling
display(order_items.head())

print('Total no of rows and columns:', order_items.shape)

display(
    order_items.count()
    .to_frame('Non null count')
    .rename_axis('Columns')
    .reset_index()
)

display(
    order_items.nunique()
    .to_frame('Unique non null counts')
    .rename_axis('Columns')
    .reset_index()
)
# Order items: 112,650 item records across 98,666 distinct orders.
# order_item_id repeats across orders; it is not a globally unique item ID.
```

```python
#Dataset 4: Order payments data profiling
display(order_payments.head())

print('Total no of rows and columns:', order_payments.shape)

display(
    order_payments.count()
    .to_frame('Non null count')
    .rename_axis('Column')
    .reset_index()
)

display(
    order_payments.nunique()
    .to_frame('Unique non null counts')
    .rename_axis('Column')
    .reset_index()
)
# Order payments: 103,886 payment records across 99,440 distinct orders.
# Some orders have multiple payment records; there are 5 payment types.
```

```python
#Dataset 5: Order reviews data profiling
display(order_reviews.head())

print('Total no of rows and columns:', order_reviews.shape)

display(
    order_reviews.count()
    .to_frame('Non null count')
    .rename_axis('Column')
    .reset_index()
)

display(
    order_reviews.nunique()
    .to_frame('Unique non null counts')
    .rename_axis('Column')
    .reset_index()
)
# Order reviews: 99,224 rows, 98,410 distinct review IDs, and 98,673 distinct order IDs.
# Both IDs repeat and require further investigation.
```

```python
#Dataset 6: Orders data profiling
display(orders.head())

print('Total no of rows and columns:', orders.shape)

display(
    orders.count()
    .to_frame('Non null counts')
    .rename_axis('Column')
    .reset_index()
)

display(orders.nunique()
        .to_frame('Unique non null counts')
        .rename_axis('Column')
        .reset_index()
)
# Orders: 99,441 rows with unique, non-null order IDs and 8 order statuses.
# Some approval and delivery timestamps are missing.
```

```python
#Dataset 7: Products data profiling
display(products.head())

print('Total no of rows and columns:', products.shape)

display(products.count()
        .to_frame('Non null count')
        .rename_axis('Column')
        .reset_index()
)

display(products.nunique()
        .to_frame('Unique non null counts')
        .rename_axis('Column')
        .reset_index()
)
# Products: 32,951 rows with unique, non-null product IDs and 73 distinct categories.
# Category names are missing for 610 products.
```

```python
#Dataset 8: Products category data profiling
display(products_category.head())

print('Total no of rows and columns:', products_category.shape)

display(
    products_category.count()
        .to_frame('Non null count')
        .rename_axis('Column')
        .reset_index()
)

display(products_category.nunique()
        .to_frame('Unique non null counts')
        .rename_axis('Column')
        .reset_index()
)
# Category translation: 71 unique Portuguese-to-English category mappings.
```

```python
#Dataset 9: Sellers data profiling
display(sellers.head())

print('Total no of rows and columns:', sellers.shape)

display(sellers.count()
        .to_frame('Non null count')
        .rename_axis('Column')
        .reset_index()
)

display(sellers.nunique()
        .to_frame('Unique non null counts')
        .rename_axis('Column')
        .reset_index()
)
# Sellers: 3,095 rows with unique, non-null seller IDs.
# Sellers span 611 cities and 23 states; multiple sellers share locations.
```
