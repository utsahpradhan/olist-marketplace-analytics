## Order Analysis

```python
%sql
SELECT
  order_status,
  COUNT(order_id) AS Order_Per_Status
FROM
  orders
GROUP BY
  order_status
ORDER BY
  Order_Per_Status DESC;
```

```python
%sql
SELECT
  min(order_purchase_timestamp) AS First_order,
  max(order_purchase_timestamp) AS Last_order
FROM
  orders;
```

```python
%sql
SELECT
  COUNT(order_id) AS Total_order,
  CAST(DATE_TRUNC('MONTH', order_purchase_timestamp) AS DATE) AS Monthly_purchase
FROM
  orders
GROUP BY
  Monthly_purchase
ORDER BY
  Monthly_purchase
```

### Delivery Performance

```python
%sql
SELECT
  ROUND(
    AVG(DATEDIFF(order_delivered_customer_date, order_purchase_timestamp)),
    2
  ) AS Avg_delivery_days
FROM
  orders
WHERE
  order_status = 'delivered'
  AND order_delivered_customer_date IS NOT NULL;
```

```python
%sql
WITH delivery_late AS (
  SELECT
    COUNT(order_id) AS Late_delivery
  FROM
    orders
  WHERE
    order_status = 'delivered'
    AND CAST(order_delivered_customer_date AS DATE) > CAST(order_estimated_delivery_date AS DATE)
),
delivery_total AS (
  SELECT
    COUNT(order_id) AS Eligible_orders
  FROM
    orders
  WHERE
    order_status = 'delivered'
    AND order_delivered_customer_date IS NOT NULL
    AND order_estimated_delivery_date IS NOT NULL
),
avg_late_delivery_days AS (
  SELECT
    ROUND(
      AVG(
        DATEDIFF(
          CAST(order_delivered_customer_date AS DATE),
          CAST(order_estimated_delivery_date AS DATE)
        )
      ),
      2
    ) AS Average_late_delivery_days
  FROM
    orders
  WHERE
    order_status = 'delivered'
    AND CAST(order_delivered_customer_date AS DATE) > CAST(order_estimated_delivery_date AS DATE)
)
SELECT
  Late_delivery,
  Eligible_orders,
  ROUND(Late_delivery * 100.0 / NULLIF(Eligible_orders, 0), 2) AS Late_delivery_percentage,
  Average_late_delivery_days
FROM
  delivery_late CROSS JOIN delivery_total CROSS JOIN avg_late_delivery_days;

--6.77% of eligible delivered orders were late.
--Late orders averaged 10.62 days past the promised date.
```

## Building the Order Summary

```python
%sql
CREATE OR REPLACE TEMP VIEW orders_with_items AS
WITH order_items_summary AS (
  SELECT
    order_id,
    ROUND(SUM(price), 2) AS Product_value,
    ROUND(SUM(freight_value), 2) AS Freight_value,
    COUNT(*) AS Item_count
  FROM
    order_items
  GROUP BY
    order_id
)
SELECT
  o.*,
  oi.Product_value,
  oi.Freight_value,
  oi.Item_count
FROM
  orders o
    LEFT JOIN order_items_summary oi
      ON o.order_id = oi.order_id;

--Preview the orders and item summary.
SELECT
  *
FROM
  orders_with_items
LIMIT 10;
```

### Adding Payment Totals

```python
%sql
CREATE OR REPLACE TEMP VIEW orders_with_items_plus_payment AS
WITH Payment_summary AS (
  SELECT
    order_id,
    ROUND(SUM(payment_value), 2) AS Payment_amount,
    count(*) AS Payment_count
  FROM
    order_payments
  GROUP BY
    order_id
)
SELECT
  owi.*,
  ps.Payment_amount,
  ps.Payment_count
FROM
  orders_with_items owi
    LEFT JOIN Payment_summary ps
      ON owi.order_id = ps.order_id;

SELECT
  *
FROM
  orders_with_items_plus_payment
limit 5
```

### Adding Review Summaries
Reviews are averaged per order before joining to preserve one row per order. The final comparison weights each reviewed order equally.

```python
%sql
CREATE OR REPLACE TEMP VIEW orders_with_items_plus_payment_plus_review AS
WITH Review_summary AS (
  SELECT
    order_id,
    ROUND(AVG(review_score), 2) AS Average_review_score,
    COUNT(*) AS Review_count
  FROM
    order_reviews
  GROUP BY
    order_id
)
SELECT
  owipp.*,
  rs.Average_review_score,
  rs.Review_count
FROM
  orders_with_items_plus_payment owipp
    LEFT JOIN Review_summary rs
      ON owipp.order_id = rs.order_id;

SELECT
  *
FROM
  orders_with_items_plus_payment_plus_review
limit 5
```

```python
%sql
SELECT
  COUNT(*) AS Total_rows,
  COUNT(DISTINCT order_id) AS Unique_orders
FROM
  orders_with_items_plus_payment_plus_review;
```

## Delivery Timing and Review Scores

```python
%sql
SELECT
  CASE
    WHEN
      CAST(order_delivered_customer_date AS DATE) <= CAST(order_estimated_delivery_date AS DATE)
    THEN
      'On time'
    ELSE 'Late'
  END AS Delivery_timing,
  COUNT(*) AS Order_Count,
  ROUND(AVG(Average_review_score), 2) AS Average_review_score
FROM
  orders_with_items_plus_payment_plus_review
WHERE
  order_status = 'delivered'
  AND order_delivered_customer_date IS NOT NULL
  AND order_estimated_delivery_date IS NOT NULL
  AND Average_review_score IS NOT NULL
GROUP BY
  Delivery_timing;

--Orders delivered late had a lower average review score (2.27) than orders delivered on time (4.29).
--This comparison shows association, not causation.
```
